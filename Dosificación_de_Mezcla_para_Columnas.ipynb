# **Dosificación de Mezcla para Columnas**

**Planteamiento:**

Un ingeniero residente necesita calcular cuántos sacos de cemento se requieren para vaciar una columna de sección circular.


 Se sabe que para un concreto estándar de $210 kg/cm^2$ (3000 psi), se consumen 9.5 sacos de cemento por cada metro cúbico de mezcla.

**Requerimientos:**

1. Solicitar la altura de la columna en metros.
2. Solicitar el diámetro de la columna en metros.
3. Calcular el volumen de la columna: $$v = \pi * r^2 * h$$ (Sugerencia: usar 3.1416
 o importar math.pi).
4. Calcular la cantidad de sacos:
$$sacos = v * 9.5$$

**Prueba del programa:**

Para una columna de 3 metros de alto y 0.4 metros de diámetro (r=0.2):

Volumen ≈0.377 m3

Sacos necesarios ≈3.58

 (el estudiante verá aquí la importancia de los tipos decimales).



````python
# ENTRADA: Solicitar la altura de la columna en metros.
# Solicitar el diámetro de la columna en metros.
h = float(input("Ingrese la altura de la columna en metros: "))
d = float(input("Ingrese el diámetro de la columna en metros: "))

# PROCESO: Calcula el radio r, calcula el volumen de la columna, calcula sacos
r = d/2
v = 3.1416 * r**2 * h
sacos = v * 9.5

# SALIDA: Imprime el resultado
print("Volumen ≈", round(v, 3), "m3. ")
print("Sacos necesarios ≈", round(sacos, 2))
````

```markdown
Aquí está la explicación línea por línea del código:

*   `h = float(input("Ingrese la altura de la columna en metros: "))`
    *   **Propósito:** Solicita al usuario que ingrese la altura de la columna en metros.
    *   **Funcionamiento:**
        *   `input(...)`: Muestra el mensaje entre comillas al usuario y espera a que escriba un valor y presione Enter.
        *   `float(...)`: Convierte la entrada del usuario (que siempre es texto) a un número decimal (flotante), ya que las medidas pueden tener decimales.
        *   `h = ...`: Almacena el valor numérico de la altura en la variable `h`.

*   `d = float(input("Ingrese el diámetro de la columna en metros: "))`
    *   **Propósito:** Similar a la línea anterior, solicita al usuario que ingrese el diámetro de la columna en metros.
    *   **Funcionamiento:** Convierte la entrada del usuario a un número decimal y la guarda en la variable `d`.

*   `r = d/2`
    *   **Propósito:** Calcula el radio de la columna.
    *   **Funcionamiento:** Divide el diámetro (`d`) por 2 y asigna el resultado a la variable `r` (radio).

*   `v = 3.1416 * r**2 * h`
    *   **Propósito:** Calcula el volumen de la columna circular.
    *   **Funcionamiento:**
        *   `3.1416`: Es una aproximación del valor de Pi (π).
        *   `r**2`: Calcula el radio (`r`) elevado al cuadrado (r * r).
        *   `*`: El operador de multiplicación.
        *   Multiplica Pi por el radio al cuadrado y por la altura (`h`) para obtener el volumen, y lo almacena en la variable `v`.

*   `sacos = v * 9.5`
    *   **Propósito:** Calcula la cantidad de sacos de cemento necesarios.
    *   **Funcionamiento:** Multiplica el volumen calculado (`v`) por la cantidad de sacos requeridos por metro cúbico (9.5) y guarda el resultado en la variable `sacos`.

*   `print("Volumen ≈", round(v, 3), "m3. ")`
    *   **Propósito:** Muestra el volumen calculado de la columna.
    *   **Funcionamiento:**
        *   `print(...)`: Es una función que imprime texto y valores en la consola.
        *   `"Volumen ≈"`: Una cadena de texto literal.
        *   `round(v, 3)`: Redondea el valor del volumen (`v`) a 3 decimales para una mejor presentación.
        *   `"m3. "`: Otra cadena de texto literal para indicar las unidades.

*   `print("Sacos necesarios ≈", round(sacos, 2))`
    *   **Propósito:** Muestra la cantidad de sacos de cemento necesarios.
    *   **Funcionamiento:** Similar a la línea anterior, imprime una cadena de texto, el valor de `sacos` redondeado a 2 decimales y una cadena de texto final.
