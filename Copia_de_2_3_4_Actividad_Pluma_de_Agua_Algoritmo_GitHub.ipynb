Aunque algunos se sorprendan o incluso se burlen del término, lo cierto es que decir “pluma” en vez de “llave” no es un error, sino una herencia viva que conecta nuestro lenguaje con siglos de evolución hidráulica.

En algunos lugares, ese flujo se regulaba mediante orificios calibrados y dispositivos llamados **plumeros**, que permitían controlar la salida del agua.

## 2. Situación problema

Un hogar recibe una cierta cantidad de **plumas de agua**.  
Se sabe que:

- **1 pluma = 0.025 litros por segundo**

Con esta información, el estudiante debe construir un programa que permita calcular:

1. el caudal total en litros por segundo;
2. el consumo diario en litros;
3. el consumo mensual en litros (suponiendo 30 días);
4. una clasificación del consumo:
   - **Bajo**: menos de 1000 litros por día
   - **Medio**: entre 1000 y 3000 litros por día
   - **Alto**: más de 3000 litros por día

Complete el siguiente bloque de código a partir del algoritmo anterior.

````python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# ==========================================================

# ENTRADA: Solicitar al usuario el número de plumas
plumas = float(input("Ingrese el número de plumas: "))

#PROCESO: Definir constantes del problema

##Definir constantes del problema

Caudal_pluma= 0.025 # litros por segundo

segundos_dia= 24 * 60 * 60 # segundos

dias_mes= 30 # dias

## Calcular el caudal total de agua que fluye a través de todas las plumas asignadas.
caudal_total= plumas * Caudal_pluma # litros/segundo

##Calcular el consumo diario

consumo_diario= caudal_total * segundos_dia # litros/dia

##Calcular el consumo mensual

consumo_mensual= consumo_diario * dias_mes # litros/mes

## Clasificar el consumo
nivel_consumo=""

if consumo_diario < 1000:
  nivel_consumo = "Bajo"
elif 1000 <= consumo_diario <= 3000:
  nivel_consumo = "Medio"
else:
  nivel_consumo = "Alto"

# SALIDA: imprimir el resultado

## Mostrar resultados
print(f" Resultados del Consumo de Agua ")
print(f"Número de plumas ingresadas: {plumas}")
print(f"Caudal total: {caudal_total:.3f} litros por segundo")
print(f"Consumo diario: {consumo_diario:.2f} litros")
print(f"Consumo mensual: {consumo_mensual:.2f} litros")
print(f"Clasificación del consumo: {nivel_consumo}")

````

### 6. Versión Comentada del Código (Sin Cambiar su Lógica)





````python

# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON

# ENTRADA
# Solicitar al usuario el número de plumas de agua


plumas = float(input("Ingrese el número de plumas: "))

# PROCESO
# Definir constantes del problema


# Cantidad de litros por segundo que consume una pluma
Caudal_pluma = 0.025

# Cantidad de segundos que tiene un día
segundos_dia = 24 * 60 * 60

# Cantidad de días considerados en un mes
dias_mes = 30


# Calcular el caudal total de agua


# Multiplica el número de plumas por el caudal de cada una
caudal_total = plumas * Caudal_pluma


# Calcular el consumo diario


# Convierte litros/segundo a litros/día
consumo_diario = caudal_total * segundos_dia

# Calcular el consumo mensual


# Multiplica el consumo diario por los días del mes
consumo_mensual = consumo_diario * dias_mes

# Clasificar el nivel de consumo


# Variable donde se guardará la clasificación
nivel_consumo = ""

# Si el consumo es menor a 1000 litros
if consumo_diario < 1000:
    nivel_consumo = "Bajo"

# Si está entre 1000 y 3000 litros
elif 1000 <= consumo_diario <= 3000:
    nivel_consumo = "Medio"

# Si supera los 3000 litros
else:
    nivel_consumo = "Alto"


# SALIDA


print(f"Resultados del Consumo de Agua")
print(f"Número de plumas ingresadas: {plumas}")
print(f"Caudal total: {caudal_total:.3f} litros por segundo")
print(f"Consumo diario: {consumo_diario:.2f} litros")
print(f"Consumo mensual: {consumo_mensual:.2f} litros")
print(f"Clasificación del consumo: {nivel_consumo}")
````
