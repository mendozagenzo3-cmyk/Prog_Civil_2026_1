# 1. Problema sencillo de ingeniería

Supongamos un caso típico en **Ingeniería Civil o Agrícola**:

**Calcular el área de un terreno rectangular.**

Este cálculo es común cuando se necesita estimar:

* área de cultivo
* área de construcción
* área de un lote

La fórmula es:

[
Área = Largo \times Ancho
]


# 2. Definir el algoritmo (Entradas – Proceso – Salidas)

## Entradas (Inputs)

Son los datos que el programa necesita.

* `largo` del terreno (metros)
* `ancho` del terreno (metros)



## Proceso

Operación matemática que se realiza:

```
area = largo * ancho
```

---

## Salidas (Outputs)

Resultados que el programa mostrará:

* Área del terreno en **m²**



# 3. Algoritmo paso a paso

1. Pedir al usuario el **largo del terreno**.
2. Pedir al usuario el **ancho del terreno**.
3. Multiplicar largo por ancho.
4. Mostrar el **área calculada**.



# 4. Pseudocódigo

Antes de programar en Python, escribimos la lógica:


Inicio

Leer largo
Leer ancho

area = largo * ancho

Mostrar "El área del terreno es:", area

Fin
```



# 5. Código en Python

```python
# Programa para calcular el área de un terreno rectangular
# Aplicación básica para ingeniería

# Entrada de datos
largo = float(input("Ingrese el largo del terreno en metros: "))
ancho = float(input("Ingrese el ancho del terreno en metros: "))

# Proceso
area = largo * ancho

# Salida
print("El área del terreno es:", area, "metros cuadrados")
```



# 6. Ejemplo de ejecución
