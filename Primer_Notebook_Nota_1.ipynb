# 1. Problema sencillo de ingeniería

Supongamos un caso típico en **Ingeniería Civil o Agrícola**:

**Calcular el área de un terreno rectangular.**

Este cálculo es común cuando se necesita estimar:

* área de cultivo
* área de construcción
* área de un lote

La fórmula es:

[
Área = Largo \times Ancho
]


# 2. Definir el algoritmo (Entradas – Proceso – Salidas)

## Entradas (Inputs)

Son los datos que el programa necesita.

* `largo` del terreno (metros)
* `ancho` del terreno (metros)



## Proceso

Operación matemática que se realiza:

```
area = largo * ancho
```



## Salidas (Outputs)

Resultados que el programa mostrará:

* Área del terreno en **m²**



# 3. Algoritmo paso a paso

1. Pedir al usuario el **largo del terreno**.
2. Pedir al usuario el **ancho del terreno**.
3. Multiplicar largo por ancho.
4. Mostrar el **área calculada**.



# 4. Pseudocódigo

Antes de programar en Python, escribimos la lógica:


Inicio

Leer ancho

area = largo * ancho

Mostrar "El área del terreno es:", area

Fin
```



# 5. Código en Python

```python
# Programa para calcular el área de un terreno rectangular
# Aplicación básica para ingeniería

```

### Un ejemplo aplicado a la Ingeniería

Imagina que necesitas calcular el **volumen de agua** necesario para llenar un reservorio cilíndrico en una finca. En lugar de darte el código final para que solo copies y pegues, analicemos la lógica que necesitaríamos:

1. Necesitamos el **radio** de la base del cilindro.
2. Necesitamos la **altura** o profundidad.
3. Aplicamos la fórmula matemática: $V = \pi \cdot r^2 \cdot h$.

Aquí tienes un ejemplo de cómo se vería esto en Python, siguiendo nuestras **buenas prácticas** de nombres descriptivos:

````python
import math

def calcular_volumen_reservorio(radio_m, altura_m):
    # Usamos la constante pi de la librería math
    volumen = math.pi * (radio_m ** 2) * altura_m
    return volumen

# Datos de entrada (Imagina que estos vienen de una medición en campo)
radio_actual = 2.5  # metros
altura_actual = 4.0  # metros

# Llamada a la función
resultado = calcular_volumen_reservorio(radio_actual, altura_actual)

print(f"El volumen del reservorio es: {resultado:.2f} metros cúbicos.")
````
