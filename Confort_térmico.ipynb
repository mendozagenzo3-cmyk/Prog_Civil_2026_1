## **Teoría del Confort Térmico en Ganado Vacuno**

El confort térmico en el ganado vacuno se refiere a un rango de condiciones ambientales en las que los animales no necesitan activar mecanismos fisiológicos o de comportamiento para mantener su temperatura corporal interna constante (homeostasis). Cuando los animales se encuentran fuera de este rango, experimentan estrés térmico, lo que puede afectar negativamente su salud, bienestar y productividad.

**Factores que influyen en el confort térmico:**

1.  **Temperatura del aire:** Es el factor más obvio, pero no el único.
2.  **Humedad relativa:** Una alta humedad reduce la capacidad del animal para disipar calor por evaporación (sudoración y jadeo).
3.  **Radiación solar:** La exposición directa al sol aumenta significativamente la carga de calor en los animales.
4.  **Velocidad del viento:** Un viento suave puede ayudar a disipar el calor, mientras que un viento fuerte en climas fríos puede aumentar la pérdida de calor.
5.  **Precipitación:** La lluvia puede mojar el pelaje, afectando el aislamiento térmico.

**Mecanismos de Termorregulación del Ganado:**

Los bovinos utilizan varios mecanismos para mantener su temperatura corporal:

*   **Disipación de calor:**
    *   **Evaporación:** Mediante la sudoración (limitada en bovinos) y el jadeo (respiración rápida y superficial).
    *   **Convección:** Transferencia de calor al aire circundante.
    *   **Radiación:** Intercambio de calor con el entorno (superficies, sol).
    *   **Conducción:** Transferencia de calor por contacto directo (e.g., con el suelo).
*   **Producción de calor:** Generado por el metabolismo de los alimentos y la actividad física.

**Impacto del Estrés Térmico:**

El estrés por calor puede provocar una serie de problemas en el ganado vacuno, incluyendo:

*   **Disminución de la producción de leche:** Es uno de los efectos más notables.
*   **Reducción del consumo de alimento:** Los animales comen menos para reducir la producción de calor metabólico.
*   **Menor ganancia de peso:** Afecta la eficiencia de la conversión alimenticia.
*   **Problemas reproductivos:** Reducción de la fertilidad, aumento de abortos.
*   **Supresión del sistema inmunológico:** Mayor susceptibilidad a enfermedades.
*   **Aumento de la mortalidad:** En casos severos.

**Índices de Confort Térmico:**

Para cuantificar el riesgo de estrés térmico, se utilizan diversos índices que combinan factores ambientales. El **Índice de Temperatura y Humedad (ITH)** es uno de los más comunes y se utiliza globalmente para evaluar el nivel de estrés térmico en el ganado. Al integrar la temperatura y la humedad, proporciona una medida más precisa que considerar solo la temperatura del aire.

# **Métodos y cálculos**
+ Índice de temperatura y humedad (producción intensiva)

$$ITH = 0.72 * (T_{bs} + T_{bh}) + 40.6$$

Donde:

1. $T_{bs}$ Temperatura de Bulbo Seco ($^oC$)
2. $T_{bh}$ Temperatura de Bulbo Húmedo ($^oC$)

Para la correlación de los datos de la especie bovina en la producción de leche se recomienda
utilizar los límites propuestos por National Weather Service:

1. Sí $ITH <= 70$ ambiente no estresante

2. Sí $ITH > 70 \space and \space ITH <= 78$ situación de alerta por estrés térmico

3. Sí $ITH > 78 \space and \space <= 83$ peligro por estrés térmico

4. Sí $ITH > 83$ situación crítica



```python
# ENTRADA: pedir tbs (temperatura de bulbo seco) y tbh (temperatura de bulbo húmedo )
tbs = float(input("Ingrese la temperatura de bulbo seco, grados centigrados: "))
tbh = float(input("Ingrese la temperatura de bulbo húmedo, grados centigrados: "))

#PROCESO: cálculo de ith (indice de temperatura y humedad)
ith = 0.72 * (tbs + tbh) + 40.6

#SALIDA: imprimir resultado de la evaluación del ith y el confort térmico
if ith <= 70:
  print("El ambiente es no estresante")
elif ith > 70 and ith <= 78:
  print("La situación es de alerta por estrés térmico")
elif ith > 78 and ith <= 83:
  print("La situación es de peligro por estrés térmico")
else:
  print("La situación es crítica")
```
